***Data Type Conversion (Data Preprocessing):*** Converts **TotalCharges** to numeric format and replaces invalid values with **NaN** for proper analysis.



```python
df['TotalCharges'] = pd.to_numeric(df['TotalCharges'], errors='coerce')
```

Purpose: Shows the first few values of the column (now in numeric form, with NaN if any invalid values were present).

```python
df['TotalCharges'].head()
```

***Purpose*** : Shows the data type of the column changed to float64 instead of object.

```python
df.dtypes
```

***Purpose:***
To ensure there are no missing values, which helps in accurate analysis and avoids errors.

```python
df['TotalCharges'] = df['TotalCharges'].fillna(df['TotalCharges'].median())
```

***Purpose*** : Removes repeated records to ensure data accuracy

```python
df.drop_duplicates(inplace=True)
```

***Feature Engineering:***
       The process of creating new meaningful variables (features) from existing data to improve analysis and model performance.


It creates a new column (Tenure_Group) in the DataFrame internally.


***Create Tenure Group***

```python
df['Tenure_Group'] = df['tenure'].apply(lambda x: 'New' if x < 12 else 'Loyal')
```

This displays the **first 5 rows** of only the selected columns (**tenure** and **Tenure_Group**) to verify how customers have been categorized.


```python
df[['tenure', 'Tenure_Group']].head()
```

***Monthly Charges Category***


Divides customers into spending groups.


```python
df['Charge_Category'] = pd.cut(df['MonthlyCharges'], bins=3, labels=['Low','Medium','High'])
```

This displays the **first 5 rows** of the selected columns to verify how **MonthlyCharges values are categorized into Low, Medium, and High groups**.


```python
df[['MonthlyCharges', 'Charge_Category']].head()
```

***Convert Senior Citizen***

This converts the SeniorCitizen column from numeric values (0, 1) into categorical values (No, Yes) for better understanding.

```python
df['SeniorCitizen'] = df['SeniorCitizen'].replace({0:'No',1:'Yes'})
```

This displays the first 5 values of the SeniorCitizen column to verify that values have been converted from 0/1 to No/Yes.

```python
df['SeniorCitizen'].head()
```

### ***Exploratory Data Analysis (EDA):***
###    The process of analyzing and summarizing data using statistics and visualizations to understand patterns, trends, and relationships.

### ***Churn Distribution*** :

Shows how many customers stayed and how many left.


***Explanation:***

This creates a bar chart showing the number of customers in each category of the Churn column (Yes / No).

***Purpose:***

To understand how many customers have left (churned) versus those who are retained.

***Insight:***

Helps identify whether churn is high or low, which is important for business decision-making.

plt.figure() → Creates a new figure and sets the size of the graph

plt.title() → Adds a title to the graph

plt.show() → Displays the graph/output on screen

```python
plt.figure(figsize=(5,3))
sns.countplot(x='Churn', data=df)
plt.title("Churn Distribution")
plt.show()
```

### ***Churn by Contract*** :

Shows how customer churn varies across different contract types.


sns.countplot(x='Contract', hue='Churn', data=df) → Creates a bar chart showing count of customers for each contract type, divided by churn (Yes/No)

plt.xticks(rotation=30) → Rotates x-axis labels for better readability

plt.show() → Displays the plot




***Insight :*** Helps identify which **contract type has higher churn**.


```python
plt.figure(figsize=(5,3))
sns.countplot(x='Contract', hue='Churn', data=df)
plt.xticks(rotation=30)
plt.show()
```

***Monthly Charges vs Churn***

Compares monthly charges of customers who left and stayed to see if higher charges lead to churn.

***Insight*** : Compares spending patterns of churned vs retained customers.

```python
plt.figure(figsize=(5,3))
sns.boxplot(x='Churn', y='MonthlyCharges', data=df)
plt.show()
```

***Tenure Analysis***

Shows how long customers stay with the company and whether shorter or longer duration customers are more likely to leave.

***Insight*** : Shows how long customers stay before leaving.

```python
plt.figure(figsize=(5,3))
sns.histplot(data=df, x='tenure', hue='Churn', bins=30)
plt.show()
```

***Payment Method Analysis***
